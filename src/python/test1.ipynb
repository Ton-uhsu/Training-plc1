# Python พื้นฐานสำหรับงาน PLC
ตัวอย่างนี้เรียงตามบทเรียน: ตัวแปร → เงื่อนไข → การวนซ้ำ → ฟังก์ชัน → โครงสร้างข้อมูล → NumPy และ Pandas  
รันทีละ Cell ด้วย **Shift + Enter**

```python
# 1) ตัวแปรและชนิดข้อมูล
machine_name = "Motor_A"  # str
temperature = 72.5           # float
production_count = 120       # int
is_running = True            # bool

print("เครื่องจักร:", machine_name)
print("อุณหภูมิ:", temperature, "°C")
print("จำนวนชิ้นงาน:", production_count)
print("กำลังทำงาน:", is_running)
```

```python
# 2) เงื่อนไข if / elif / else
temperature = 82

if temperature > 80:
    status = "ALARM: อุณหภูมิสูงเกินไป"
elif temperature > 60:
    status = "WARNING: ควรเฝ้าระวัง"
else:
    status = "NORMAL: อุณหภูมิปกติ"

print(status)
```

```python
# 3) for: อ่านค่าเซนเซอร์หลายจุด
sensor_values = [25.1, 26.3, 24.8, 27.0]

for sensor_number, value in enumerate(sensor_values, start=1):
    print(f"Sensor {sensor_number}: {value} °C")
```

```python
# 4) while: จำลองการนับรอบการผลิต
cycle = 1

while cycle <= 3:
    print(f"กำลังทำงานรอบที่ {cycle}")
    cycle += 1

print("จบการทำงาน")
```

```python
# 5) ฟังก์ชัน: แปลงองศาเซลเซียสเป็นฟาเรนไฮต์
def celsius_to_fahrenheit(celsius):
    return celsius * 9 / 5 + 32

temperature_c = 25
temperature_f = celsius_to_fahrenheit(temperature_c)
print(f"{temperature_c} °C = {temperature_f} °F")
```

```python
# 6) List และ Dictionary
temperatures = [25.1, 26.3, 24.8]
temperatures.append(27.0)

machine = {
    "id": "M01",
    "temperature": temperatures[-1],
    "running": True,
}

print("อุณหภูมิทั้งหมด:", temperatures)
print("ข้อมูลเครื่องจักร:", machine)
```

```python
# 7) Tuple และ Set
machine_position = (10, 20)
alarm_codes = {101, 102, 101, 103}

print("ตำแหน่งเครื่องจักร:", machine_position)
print("รหัส Alarm ที่ไม่ซ้ำ:", alarm_codes)
```

```python
# 8) NumPy: คำนวณข้อมูลเซนเซอร์
import numpy as np

temperatures = np.array([25.1, 26.3, 24.8, 27.0])
print("ค่าเฉลี่ย:", temperatures.mean())
print("ค่าสูงสุด:", temperatures.max())
print("อุณหภูมิ °F:", temperatures * 1.8 + 32)
```

```python
# 9) Pandas: สร้างตารางข้อมูล PLC โดยไม่ต้องใช้ไฟล์ CSV
import pandas as pd

plc_data = pd.DataFrame({
    "machine": ["M01", "M01", "M01", "M01"],
    "temperature": [25.1, 26.3, 24.8, 82.0],
    "motor_running": [True, True, True, False],
})

plc_data["status"] = np.where(
    plc_data["temperature"] > 80, "ALARM", "NORMAL"
)
plc_data
```
